```python
import numpy as np
import pandas as pd
```

```python
def standardize(X):
    mean = np.mean(X, axis=0)
    std = np.std(X, axis=0)
    return (X - mean) / std
```

```python
def compute_covariance(X):
    return np.cov(X.T)
```

```python
def eigen_decomposition(cov_matrix):
    eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)
    # Sort by descending eigenvalue
    idx = np.argsort(eigenvalues)[::-1]
    return eigenvalues[idx], eigenvectors[:, idx]
```

```python
def project_data(X, eigenvectors, k):
    return np.dot(X, eigenvectors[:, :k])
```

```python
def pca(X, k):
    X_std = standardize(X)
    cov_matrix = compute_covariance(X_std)
    eigenvalues, eigenvectors = eigen_decomposition(cov_matrix)
    X_pca = project_data(X_std, eigenvectors, k)
    return X_pca, eigenvalues[:k], eigenvectors[:, :k]
```

```python
# Testin with random data from Copilot
X = np.array([[2.5, 2.4],
              [0.5, 0.7],
              [2.2, 2.9],
              [1.9, 2.2],
              [3.1, 3.0],
              [2.3, 2.7],
              [2, 1.6],
              [1, 1.1],
              [1.5, 1.6],
              [1.1, 0.9]])

X_pca, eigvals, eigvecs = pca(X, k=1)
print("Projected Data:\n", X_pca)
```
